```python
import pandas as pd

# Load data
file_path = "../data/scraping/genre/artist_genres_clean.csv"
df = pd.read_csv(file_path)

# Original genre columns
genre_cols = [f'genre_{i}' for i in range(1, 14)]

# Example 7-category mapping (expand later)
genre_mapping = {
    "pop": "Pop", "synthpop": "Pop", "teen pop": "Pop",
    "dance pop": "Pop", "indie pop": "Pop", 
    "rock": "Rock", "punk rock": "Rock", "alternative rock": "Rock",
    "classic rock": "Rock", "metal": "Rock", "hard rock": "Rock",
    "hip hop": "Hip-Hop/Rap", "rap": "Hip-Hop/Rap", "trap": "Hip-Hop/Rap",
    "gangsta rap": "Hip-Hop/Rap", 
    "edm": "Electronic/Dance", "techno": "Electronic/Dance",
    "house": "Electronic/Dance", "dubstep": "Electronic/Dance",
    "jazz": "Jazz/Blues", "smooth jazz": "Jazz/Blues",
    "blues": "Jazz/Blues",
    "classical": "Classical", "baroque": "Classical", "romantic": "Classical",
    "folk": "World/Other", "latin": "World/Other", "country": "World/Other",
    "reggae": "World/Other", "afrobeat": "World/Other"
}

# Map to 7 categories but keep original genres intact
def map_to_categories(row):
    mapped = set()
    for col in genre_cols:
        genre = str(row[col]).lower() if pd.notnull(row[col]) else None
        if genre and genre in genre_mapping:
            mapped.add(genre_mapping[genre])
        elif genre and genre != "nan":
            mapped.add("Other/Uncategorized")
    return sorted(mapped)

df["categories"] = df.apply(map_to_categories, axis=1)

# Save
df.to_csv("../data/scraping/genre/artist_genres_with_categories.csv", index=False)
```

```python
# Collect all unique genres
all_genres = set()
for col in genre_cols:
    all_genres.update(df[col].dropna().unique())

print(f"Number of unique genres: {len(all_genres)}")
print(sorted(all_genres))  # Sorted list of unique genres

df_unique_genres=pd.DataFrame(all_genres)
# Save
df_unique_genres.to_csv("../data/scraping/genre/unique_genres.csv", index=False)
```
